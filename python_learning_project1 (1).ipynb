```python
# @title
 random

def get_choices():
  player_choice = input("Enter a choice (rock, paper, scissors): ").lower()
  options = ["rock", "paper", "scissors"]
  if player_choice in options:
    computer_choice = random.choice(options)
    choices = {"player": player_choice, "computer": computer_choice}
    return choices
  else:
    print("Invalid choice. Please choose rock, paper, or scissors.")
    return get_choices() # Ask again for a valid choice

def check_win(player, computer):
  print(f"You chose {player}, computer chose {computer}")
  if player == computer:
    return "It's a tie!"
 import elif (player == "rock" and computer == "scissors") or \
       (player == "paper" and computer == "rock") or \
       (player == "scissors" and computer == "paper"):
    return "You win!"
  else:
    return "Computer wins!"

# Game loop to play Rock-Paper-Scissors
while True:
  choices = get_choices()
  if choices: # Ensure valid choices were returned
    result = check_win(choices["player"], choices["computer"])
    print(result)
    play_again = input("Play again? (yes/no): ").lower()
    if play_again != 'yes':
      break
  else:
    # This case should ideally not be reached with the current get_choices logic
    # but it's here for robustness in case get_choices is modified later
    print("Could not get valid choices. Exiting game.")
    break
```

```python
!pip install -q openai

from openai import OpenAI
import getpass
import base64
from IPython.display import display
from PIL import Image

# Enter your OpenAI API key
api_key = getpass.getpass("Enter your OpenAI API key: ")
client = OpenAI(api_key=api_key)

# Prompt for realistic Lord Venkateswara
prompt = """
Create a highly realistic and respectful devotional depiction of
Lord Venkateswara of Tirumala.

Lord Venkateswara is standing majestically inside a grand traditional
Tirumala-style temple sanctum. Show the iconic divine black-stone-like
appearance, traditional silk garments, ornate golden crown, elaborate
gold jewelry, sacred Vaishnavite markings, and beautiful traditional
decorations.

Use extremely realistic textures and details, warm golden temple
lighting, natural shadows, intricate jewelry, a peaceful divine
atmosphere, and a majestic sacred presence.

Photorealistic devotional artwork, ultra-detailed, cinematic lighting,
high resolution, realistic temple environment, respectful Hindu
religious representation.
"""

# Generate image
result = client.images.generate(
    model="gpt-image-1",
    prompt=prompt,
    size="1024x1024"
)

# Save image
image_bytes = base64.b64decode(result.data[0].b64_json)

with open("Lord_Venkateswara.png", "wb") as f:
    f.write(image_bytes)

# Display image
print("🙏 Lord Venkateswara image generated successfully!")
display(Image.open("Lord_Venkateswara.png"))
```

```python
# ============================================
# LORD VENKATESWARA AI IMAGE GENERATOR
# Google Colab - No OpenAI API Key Required
# ============================================

# 1. Install required libraries
!pip install -q diffusers transformers accelerate safetensors

# 2. Import libraries
import torch
from diffusers import StableDiffusionPipeline
from IPython.display import display
from PIL import Image

# 3. Check GPU
if torch.cuda.is_available():
    device = "cuda"
    print("✅ GPU detected!")
else:
    device = "cpu"
    print("⚠️ GPU not detected. Generation will be very slow.")

# 4. Load the AI image-generation model
model_id = "runwayml/stable-diffusion-v1-5"

pipe = StableDiffusionPipeline.from_pretrained(
    model_id,
    torch_dtype=torch.float16 if device == "cuda" else torch.float32,
    safety_checker=None
)

pipe = pipe.to(device)

# 5. Prompt
prompt = """
A highly realistic and respectful devotional depiction of Lord Venkateswara
of Tirumala, standing majestically inside a beautiful traditional Indian
temple sanctum.

Lord Venkateswara has a divine dark stone-like appearance, an elaborate
golden crown, traditional silk garments, magnificent golden jewelry,
sacred Vaishnavite markings, detailed traditional decorations and a
peaceful divine expression.

Grand temple background, warm golden lamp lighting, realistic shadows,
intricate jewelry details, highly detailed textures, realistic
photography, cinematic lighting, photorealistic devotional artwork,
majestic and sacred atmosphere, ultra detailed, high resolution.
"""

# 6. Negative prompt
negative_prompt = """
cartoon, anime, blurry, low quality, distorted face, extra arms,
extra hands, extra fingers, deformed body, duplicate person,
bad anatomy, text, watermark, logo
"""

# 7. Generate image
print("🙏 Generating Lord Venkateswara... Please wait.")

result = pipe(
    prompt=prompt,
    negative_prompt=negative_prompt,
    num_inference_steps=30,
    guidance_scale=7.5
)

image = result.images[0]

# 8. Save image
image.save("Lord_Venkateswara.png")

# 9. Display image
print("✅ Image generated successfully! 🙏")
display(image)

# 10. Download the image
from google.colab import files
files.download("Lord_Venkateswara.png")
```
